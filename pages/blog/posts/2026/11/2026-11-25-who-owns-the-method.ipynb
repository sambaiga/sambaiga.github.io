The quickest way to watch a high-performing data science team disengage is to hand it a ticket that reads: "Build a random forest on these five features to predict churn by Friday."

In my experience of leading and working beside data science teams, a lack of technical talent or bad infrastructure is not the most persistent failure mode. A leadership habit is: dictating the why, the what and the how all at once.

A brief of that kind settles the method before the team has looked at the data. The team receives a specification, and its members are reduced from strategic problem-solvers to ticket-takers. A team that may only execute someone else's prescription cannot do the one thing it was hired for, which is to discover what the data is saying.

The argument of this post is an inference from experience and from the books cited, not a tested result. The section on limits states where it stops.


## Leader's Habit

Most leaders of data teams were promoted because they were strong individual contributors, and the skill that earned the promotion is the hardest to put down. S. Eblin organises each chapter of *The Next Level* around something to pick up and something to let go.[^eblin] Applied to a data team, the leader picks up the work of giving context, removing obstacles in the organisation and setting direction, and lets go of choosing the model.

A question settles most cases. Is a stated method a constraint of the business, such as latency, privacy or budget, or is it the method that the leader would have chosen as an individual contributor? A constraint stays in the brief, and a preference does not.

T. Fadell draws a related line between decisions that rest on opinion and decisions that rest on data, and describes the first version of a product as led by the opinion of its maker and later versions as led by evidence.[^fadell] I read a data science project as closer to the later case, because the evidence is in the hands of the team and not of the leader.

[^eblin]: S. Eblin, *The Next Level: What Insiders Know About Executive Success*, 3rd ed. London, U.K.: Nicholas Brealey, 2018.

[^fadell]: T. Fadell, *Build: An Unorthodox Guide to Making Things Worth Making*. New York, NY, USA: Harper Business, 2022.


## Cost of a Prescription

The cost of a prescribed method is specific to data science, and it has two sources.

**A signal is not known in advance.** Whether a usable signal exists in the data is rarely known before the exploration, so a method chosen in advance is a guess made by the person furthest from the data.

**Knowledge sits near the data.** F. A. Hayek made the general point in 1945: the knowledge that a good decision needs is the knowledge of particular circumstances of time and place, which is held by the people who are present and not by the people who command.[^hayek] The data scientists who work with the pipelines, the missing values and the anomalies hold that knowledge. A manager in a meeting room holds a summary of it.

[^hayek]: F. A. Hayek, "The use of knowledge in society," *Am. Econ. Rev.*, vol. 35, no. 4, pp. 519-530, 1945.


## Commitment and Ownership

A prescription also costs the team its commitment. P. Lencioni places commitment and accountability in the middle of the dysfunctions of a team: people commit to a decision when they have been heard on it, even if it is not the decision they would have taken, and a team that has not committed does not hold itself accountable for the result.[^lencioni] A prescribed method removes the hearing. The team may deliver what was specified, but it has no reason to defend the specification or to report that the method does not work.

Research on empowerment points the same way. A meta-analysis of studies of psychological empowerment, which includes the sense of autonomy and of influence over the work, links it to higher task performance and to higher job satisfaction.[^seibert]

[^lencioni]: P. Lencioni, *The Five Dysfunctions of a Team: A Leadership Fable*. San Francisco, CA, USA: Jossey-Bass, 2002.

[^seibert]: S. E. Seibert, G. Wang and S. H. Courtright, "Antecedents and consequences of psychological and team empowerment in organizations: A meta-analytic review," *J. Appl. Psychol.*, vol. 96, no. 5, pp. 981-1003, 2011.


## Division of Responsibility

Three questions organise a project: why it matters, which outcome it should move, and how the outcome is reached. Leadership owns the first two together with product. The data science team owns the third.

The product operating model draws the same line. M. Cagan separates teams that are given a roadmap of solutions and measured on output from teams that are given a problem and measured on the outcome, and argues that the second kind produces the results of value.[^cagan] The books on leading under uncertainty divide the work in the same way. S. McChrystal and colleagues describe shared consciousness, in which everyone understands the situation and the aim, together with empowered execution, in which the people closest to the work decide how to act.[^mcchrystal] L. D. Marquet summarises the practice as giving control to the team while creating clarity about the aim.[^marquet] D. H. Pink lists technique, the way the work is done, among the four kinds of autonomy that sustain motivation, with task, time and team.[^pink]

@fig-ownership shows the division for one project briefed in two ways. Under the prescriptive brief, leadership owns the reason, the outcome and the method. Under the problem brief, leadership and product own the reason and the outcome, for example a 5% reduction of the 30-day churn rate, and the data science team owns the method.

The method is where a team that owns it can find that a model is not the right tool, and that a cohort analysis, a simple rule or a repair of the upstream logging may answer the question at a lower cost. A team that owns the method will say so. A team that holds only a specification cannot.

The same division applies to strategy. The post on [the strategic kernel and the AI strategy gap](/pages/blog/posts/2026/03/index.html) argues that the reason, the outcome and the capability of an AI programme need to be stated together. A brief written as a problem is the form that this takes for one team.

## Case for Direction

Not every brief should be open. A. S. Grove ties the style of a manager to the task-relevant maturity of the person managed: a person who is new to a task needs structure and close direction, and a person who is experienced in it needs the objective and the freedom to choose the route.[^grove] A data scientist in the first months, an incident that must be closed within hours, and a method fixed by regulation or by safety all call for specification.

Autonomy also has a cost where a team does not check itself. Langfred found, in self-managing teams, that high trust combined with high individual autonomy lowered the performance of the team when its members did not monitor one another's work.[^langfred] The brief therefore fixes the constraints and the points of review, and the method stays with the team.

[^grove]: A. S. Grove, *High Output Management*. New York, NY, USA: Random House, 1983.

[^langfred]: C. W. Langfred, "Too much of a good thing? Negative effects of high trust and individual autonomy in self-managing teams," *Acad. Manage. J.*, vol. 47, no. 3, pp. 385-399, 2004.


## Practical Changes

Three changes follow for a leader who recognises the habit.

**Write the brief as a problem, not as a specification.** In place of "build a predictive model for churn", the brief states what is difficult to understand or to reduce, why it matters now, the constraints and the measure of success, and asks the team how it would investigate.

**Ask whether each method is a constraint or a preference.** The question comes before every kickoff, and only the constraints stay in the brief.

**Give the team a forum for the problems it finds.** Protected time and a regular meeting in which data scientists bring problems that they found in the data let the team originate work as well as receive it. The share of capacity for that time is a local decision.

A leader who stops telling capable people how to do their jobs and trusts them with hard problems gets more than better models. The team thinks and acts like an owner.


## Limits of the Argument

The post rests on experience and on books written for leaders in general, not on a study of data science teams. The empirical source shows that empowerment is associated with performance, and it does not show that a prescribed method causes a team to disengage. The claim that ownership of the method improves the work is a proposition that a leader can test in one team, by comparing a project briefed as a problem with one briefed as a specification.
