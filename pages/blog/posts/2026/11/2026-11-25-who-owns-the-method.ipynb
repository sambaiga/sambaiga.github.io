Most people who have worked in an organisation have received an instruction that stated what the result should be and also how to obtain it. A leader who stood further from the work than the person who carried it out had written it. Such an instruction is usually well meant, since it saves a discussion and reflects a route that has worked before. Its effect on a capable team is nonetheless predictable. The team delivers what was specified and stops reporting what it sees, including the point at which the specified route fails. In a data science team the effect is stronger, because the route is the part of the work that is least known in advance. A brief that reads "build a random forest on these five features to predict churn by Friday" fixes the model, the inputs and the date before anyone has looked at the data.

In my experience of leading and working beside data science teams, this habit is a more persistent cause of disengagement than a shortage of technical skill or a weak infrastructure. The reason for the work, the outcome that it should achieve and the method that produces it are all settled at once, before the team has contributed. A team that may only carry out the prescription of someone else cannot do what it was hired to do, which is to find out what the data is able to say. Its members gradually cease to act as problem-solvers.

This post argues that the method should belong to the team, and that leadership should hold the reason and the outcome together with the constraints of the business. The argument rests on experience, on a small number of books on leadership and on three studies. It is not a measured result, and its limits are stated in the last section. The post also continues three earlier ones, on [shared mental models](/pages/blog/posts/2024/07/index.html), on [group forces in organisational change](/pages/blog/posts/2024/09/index.html) and on [the strategic kernel of an AI strategy](/pages/blog/posts/2026/03/index.html). They supply the neighbouring arguments that are referred to below.


## Leadership Tendency to Prescribe

A leader who can see a route to an answer is tempted to settle the whole of it: why the problem matters, which outcome to pursue and how to reach it. The tendency is natural, because the route looks clear from the position of the leader, and it is the habit that S. Eblin asks leaders to change. Each chapter of *The Next Level* begins with something to pick up and something to let go.[^eblin] Applied to a data team, the leader picks up the work of giving context, removing obstacles in the organisation and setting direction, and lets go of the choice of the model. The chapter titles of the first edition indicate the direction of the advice: team reliance, defining what to do, and accountability for many results.[^eblinfirst] I read the three together as a division of labour. The leader relies on the team, defines what is to be done and holds the team accountable for the result, but does not carry out the work of the team in its place.

Eblin also asks leaders to watch the dance from the balcony.[^eblin]<!-- confirm the chapter and page of the balcony passage in the 3rd edition --> On the dance floor a leader sees the next step and the partner in front. From the balcony the leader sees the whole floor, including who is dancing with whom and where the floor is crowded or empty. @fig-balcony draws the image. The walls of the floor are the constraints of the business, the paths of the dancers are the method, and only the balcony shows a corner that no dancer has used. A leader who stays on the floor chooses the next step for the team. A leader on the balcony asks what the floor requires and leaves the steps to the dancers. The image supplies a test for each method that appears in a brief. A method may be a constraint of the business, such as latency, privacy or budget. It then describes the floor and stays in the brief. Otherwise it is the step that the leader would have chosen personally, which makes it a preference that passes to the dancers.

T. Fadell draws a related line between decisions that rest on opinion and decisions that rest on data. He describes the first version of a product as led by the opinion of its maker and later versions as led by evidence.[^fadell] I read a data science project as closer to the later case, because the evidence lies in the hands of the team and not of the leader.

## Cost of a Prescribed Method

The cost of a prescribed method has two sources that are specific to data science. The first is that a usable signal in the data is rarely known to exist before the exploration. A method chosen in advance is therefore a guess made by the person who is furthest from the data. The second is that the knowledge needed to choose well sits close to the data. F. A. Hayek made the general point in 1945. The knowledge on which a good decision depends, in his account, is knowledge of the particular circumstances of time and place. It is held by the people who are present and not by those who command.[^hayek] The data scientists who work with the pipelines, the missing values and the anomalies hold that knowledge, and a manager in a meeting room holds a summary of it.

@fig-knowledge sets out what each side holds. Leadership holds the purpose of the work and the constraints of the business, and the team holds the pipelines, the anomalies and the signal in the data. The brief is the channel between them. Context and constraints pass to the team through the brief. The findings and the risks that the team sees need a channel in the opposite direction, which a leader who has already fixed the method has little reason to open.

[^hayek]: F. A. Hayek, "The use of knowledge in society," *Am. Econ. Rev.*, vol. 35, no. 4, pp. 519-530, 1945.

## Commitment and Ownership

A prescription also costs the team its commitment. P. Lencioni places commitment and accountability in the middle of the dysfunctions of a team. People commit to a decision when they have been heard on it, even if it is not the decision that they would have taken. A team that has not committed does not hold itself accountable for the result.[^lencioni] A prescribed method removes the hearing. The team may deliver what was specified, but it has no reason to defend the specification or to report that the method does not work. The leader therefore learns of a failure late.

The same pattern appears in the account of organisational change in an earlier post on Lewin's theory. Lewin's method of action research asks a group to analyse the situation, to identify alternatives, and to choose and act. Change is then made with the people concerned and not to them.[^burnes] A brief that gives a data science team the problem and its constraints invites the team into the same cycle. A prescribed method has already completed the first two steps on the team's behalf. A second earlier post, on shared mental models, describes the common understanding of goals, roles and tasks that distinguishes effective teams. It reports that teams in which leadership is shared show greater commitment and more sharing of information.[^lungeanu] A brief written as a problem builds that understanding of the goal in the open, because the team and its leaders discuss the aim and the constraints before the method is chosen.

Research on empowerment points the same way. A meta-analysis of studies of psychological empowerment, which includes the sense of autonomy and of influence over the work, links it to higher task performance and to higher job satisfaction.[^seibert] @fig-commitment follows the chain from being heard to the result, in a form simplified from Lencioni. Under a prescribed method the team is not heard and the later links are not reached, whereas under a problem brief the chain runs to the result.

## Division of Responsibility

The preceding sections lead to a division of the three questions that organise a project: why the work matters, which outcome it should move and how the outcome is reached. Leadership answers the first two together with product, and the data science team answers the third. The division works because each side holds what the other cannot see: leadership sees the purpose of the work and the limits of the business, and the team sees the data. A leader who states the aim does not need to choose the method, and a team that understands the aim is able to choose it. M. Cagan draws the same line between product teams and feature teams. The first is given a problem and measured on the outcome, and it chooses the solution. The second is given a roadmap and measured on output, and it builds a solution that was chosen for it.[^cagan]

Two accounts of leading under pressure show the condition that makes the division safe. S. McChrystal describes a task force in Iraq that paired shared consciousness with empowered execution. Every unit understood the situation and the aim, and the people closest to the action decided how to act.[^mcchrystal] L. D. Marquet, who commanded a nuclear submarine, summarises his practice as giving control to the crew while creating clarity about the aim.[^marquet] The condition in both accounts is clarity, since control without a shared aim leads to drift and clarity without control amounts to a command. A brief written as a problem supplies the clarity, and the method supplies the control. The motivation of the team points the same way. D. H. Pink lists technique, the way the work is done, among the four kinds of autonomy that sustain motivation, with task, time and team,[^pink] and a brief that fixes the method removes one of the four.

@fig-ownership shows the division for one project briefed in two ways. Under the prescriptive brief, leadership owns the reason, the outcome and the method. Under the problem brief, leadership and product own the reason and the outcome, for example a 5% reduction of the 30-day churn rate, and the data science team owns the method. The method is where a team that owns it can find that a model is not the right tool. A cohort analysis, a simple rule or a repair of the upstream logging may answer the question at a lower cost, and a team that holds only a specification has no occasion to report that.

The division also restates, for one project, the layers of the strategic kernel that the earlier posts on [the kernel](/pages/blog/posts/2026/03/index.html) and on [a five-step roadmapping process](/pages/blog/posts/2026/04/index.html) describe. The diagnosis, which those posts call Know-Why, and the guiding policy, which they call Know-What, are the work of leadership. The coherent actions, which they call Know-How, rest on the skills and the capabilities of the people who carry them out. The roadmap asks that every initiative connect back to the crux that the diagnosis identified. A brief written as a problem keeps that connection, because the team receives the crux and the constraints and chooses the action.

## Conditions for Specifying the Method

Not every brief should be open. A. S. Grove ties the style of a manager to the task-relevant maturity of the person managed: a person who is new to a task needs structure and close direction, and a person who is experienced in it needs the objective and the freedom to choose the route.[^grove] A data scientist in the first months, an incident that must be closed within hours, and a method fixed by regulation or by safety all call for specification. @fig-maturity sketches the relation as a conceptual illustration and not as data. Direction falls as the maturity of the person for the task rises, regulation, safety or urgency raise it at any maturity, and the review points stay at every level.

Autonomy also has a cost where a team does not check itself. Langfred found, in self-managing teams, that high trust combined with high individual autonomy lowered the performance of the team when its members did not monitor one another's work.[^langfred] The brief therefore fixes the constraints and the points of review, and the method stays with the team.

## Changes in Leadership Practice

Three changes follow for a leader who recognises the habit, and @fig-changes summarises them. The first is to write the brief as a problem and not as a specification. In place of "build a predictive model for churn", the brief states what is difficult to understand or to reduce, why it matters now, the constraints and the measure of success, and asks the team how it would investigate. The second is to put each method that appears in a brief to the test of the balcony, so that only the constraints remain and the preferences pass to the team. The third is to give the team protected time and a regular forum in which data scientists present the problems that they found in the data. The team then originates work as well as receives it, and the share of capacity for that time is a local decision.

A leader who gives the team a hard problem and its constraints, and then trusts the team with the method, gains more than a better model. The team begins to act as the owner of the work and reports what it finds, whether or not it was asked.

## Limits of the Argument

The post rests on experience and on books written for leaders in general, not on a study of data science teams. The empirical sources show that empowerment is associated with performance and that shared leadership is associated with commitment, and they do not show that a prescribed method causes a team to disengage. A leader can test the claim that ownership of the method improves the work in one team, by comparing a project briefed as a problem with one briefed as a specification.
